## 10.1.1 Iterating over Groups

```python
for name, group in df.groupby("key1"):
    print(name)
    print(group)
```

```python
for (k1, k2), group in df.groupby(["key1", "key2"]):
    print((k1, k2))
    print(group)
```

```python
grouped = df.groupby({"key1":"key", "key2":"key",
                      "data1": "data", "data2": "data"}, axis = "columns")
```

```python
for group_key, group_values in grouped:
    print(group_key)
    print(group_values)
```

```python
def peak_to_peak(arr):
    return arr.max() - arr.min()
```

### 10.2.1 Column-Wise and Multiple Function Application

```python
tips = pd.read_csv("examples/tips.csv")
```

```python
tips.head(10)
```

```python
tips["tip_pct"] = tips["tip"] / tips["total_bill"]
```

```python
tips.head()
```

```python
grouped = tips.groupby(["day", "smoker"])
```

```python
grouped_pct = grouped["tip_pct"]
```

```python
grouped_pct.agg("mean")
```

```python
grouped_pct.agg(["mean", "std", peak_to_peak])
```

```python
grouped_pct.agg([("average", "mean"), ("stdev", np.std)])
```

```python
functions = ["count", "mean", "max"]
```

```python
result = grouped[["tip_pct", "total_bill"]].agg(functions)
```

```python
result
```

```python
result["tip_pct"]
```

```python
ftuples = [("Average", "mean"), ("Variables", np.var)]
```

```python
grouped[["tip_pct", "total_bill"]].agg(ftuples)
```

```python
grouped.agg({"tip": np.max, "size": "sum"})
```

```python
grouped.agg({"tip_pct": ["min", "max", "mean", "std"], 
             "size": "sum"})
```

```python
tips.groupby(["day", "smoker"], as_index=False).mean(numeric_only = True)
```

```python
def top(df, n=5, column="tip_pct"):
    return df.sort_values(column, ascending = False)[:n]
```

```python
top(tips, n=6)
```

```python
tips.groupby("smoker").apply(top)
```

```python
tips.groupby(["smoker", "day"]).apply(top, n=1, column = "total_bill")
```

```python
result = tips.groupby("smoker")["tip_pct"].describe()
```

```python
result
```

```python
result.unstack("smoker")
```

```python
tips.groupby("smoker", group_keys =False).apply(top)
```

```python
frame = pd.DataFrame({"data1": np.random.standard_normal(1000), 
                      "data2": np.random.standard_normal(1000)})
```

```python
frame.head()
```

```python
quartiles = pd.cut(frame["data1"], 4)
```

```python
quartiles.head(10)
```

```python
def get_stats(group):
    return pd.DataFrame(
        {
            "min": group.min(), 
            "max": group.max(), 
            "count": group.count(), 
            "mean": group.mean()
        }
    )
```

```python
grouped = frame.groupby(quartiles)
```

```python
grouped.apply(get_stats)
```

```python
grouped.agg(["min", "max", "count", "mean"])
```

```python
quartiles_samp = pd.qcut(frame["data1"], 4, labels = False)
```

```python
quartiles_samp.head(10)
```

```python
grouped.apply(get_stats)
```

```python
states = ["Ohio", "New York", "Vermont", "Florida",
          "Oregon", "Nevada", "California", "Idaho"]
```

```python
group_key = ["East", "East", "East", "East",
             "West", "West", "West", "West"]
```

```python
data = pd.Series(np.random.standard_normal(8), index = states)
```

```python
data
```

```python
data[["Vermont", "Nevada", "Idaho"]] = np.nan
```

```python
data
```

```python
data.groupby(group_key).size()
```

```python
data.groupby(group_key).count()
```

```python
data.groupby(group_key).mean()
```

```python
def fill_mean(group):
    return group.fillna(group.mean())
```

```python
data.groupby(group_key).apply(fill_mean)
```

```python
fill_values = {"East": 0.5, "West": -1}
```

```python
def fill_func(group):
    return group.fillna(fill_values[group.name])
```

```python
data.groupby(group_key).apply(fill_func)
```

```python
suits = ["H", "S", "C", "D"]  # Hearts, Spades, Clubs, Diamonds（红桃，黑桃，梅花，方块）
card_val = (list(range(1, 11)) + [10] * 3) * 4
base_names = ["A"] + list(range(2, 11)) + ["J", "K", "Q"]
cards = []
for suit in suits:
    cards.extend(str(num) + suit for num in base_names)

deck = pd.Series(card_val, index=cards)
```

```python
deck.head(13)
```

```python
def draw(deck, n=5):
    return deck.sample(n)
```

```python
draw(deck)
```

```python
def get_suit(card):
    #最后一个字母是花色
    return card[-1]
```

```python
deck.groupby(get_suit).apply(draw, n=2)
```

```python
deck.groupby(get_suit, group_keys = False).apply(draw, n=2)
```
