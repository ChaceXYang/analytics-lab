```python
data = pd.DataFrame(np.arange(16).reshape((4, 4)), 
                   index = ["Ohio", "Colorado", "Utah", "New York"], 
                   columns = ["one", "two", "three", "four"])
```

```python
data[data < 5] = 0
```

```python
data
```

```python
data["two"] #取一列 -> Sereies
```

```python
data[["one", "two"]] #取多列 -> DataFrame
```

```python
data.loc[:, "two"] #取two列
```

```python
data.loc[:, ["one", "two"]] # 取one列和two列
```

```python
data.loc["Colorado":"Utah", "two":"three"] 
```

```python
data.iloc[0] #取第0行 -> Ohio哪一行
```

```python
data.iloc[1:3] # 取第1到2行（不含第三行）
```
