```python
import pandas as pd
```

```python
import numpy as np
```

```python
import sqlite3
```

```python
train = pd.read_csv(
    "data/transactions_train.csv",
    usecols=['t_dat', 'customer_id', 'article_id', 'price'], 
    chunksize = 2000000
)
```

```python
chunks = []                    # 空「列表」，不是 DataFrame
```

```python
for chunk in train:            # 遍历那个带 chunksize 的迭代器
    chunk['month'] = chunk['t_dat'].str[:7]
    grouped = chunk.groupby(['customer_id', 'month']).agg(
        active_days = ('t_dat', 'nunique'), # 当月有交易的天数
        last_date = ('t_dat', 'max'), #当月最后一次交易日期
        amount = ('price', 'sum'), # 当月总消费额
        items = ('article_id', 'count') # 当月购买件数
    ).reset_index()
    chunks.append(grouped)     # 每块结果塞进列表
```

```python
monthly = pd.concat(chunks, ignore_index=True)   # 循环结束一次性拼成
```

```python
monthly.shape
```

```python
monthly.head(100)
```

```python
monthly.duplicated(subset=['customer_id', 'month']).sum()
```

```python
monthly = monthly.groupby(['customer_id', 'month']).agg(
    active_days = ('active_days', 'sum'), # 当月有交易的天数
    last_date = ('last_date', 'max'), #当月最后一次交易日期
    amount = ('amount', 'sum'), # 当月总消费额
    items = ('items', 'sum') # 当月购买件数
)
```

```python
monthly = monthly.reset_index()
```

```python
age = pd.read_csv("data/customers.csv", usecols=['customer_id', 'age', 'club_member_status'])
```

```python
age
```

```python
def clean_age(a):
    if pd.isna(a):
        return np.nan
    elif a < 16 or a > 90:
        return np.nan
    else:
        return a
```

```python
age['age_clean'] = age['age'].apply(clean_age)
```

```python
bins=[15, 25, 35, 45, 55, 90]
```

```python
labels=['16-25', '26-35', '36-45', '46-55', '56+']
```

```python
age['age_band'] = pd.cut(age['age_clean'],
                         bins=[15, 25, 35, 45, 55, 90],
                         labels=['16-25', '26-35', '36-45', '46-55', '56+'])
```

```python
age.head()
```

```python
age['age_clean'].isna().mean() #缺失占比
```

验证清洗"生效了、且不过火"。
如果是 0，说明没清到东西（但数据里明明有 100 多岁的人，不可能是 0）；
如果冲到 30%，说明你规则太狠、误伤正常值了。
1.16% 这个量级是合理的。

```python
age['age_band'].value_counts() #每个年龄段有多少人
```

```python
# 连接数据库（文件落盘，下次不用重跑 3.7GB）
conn = sqlite3.connect('data/hm.db')
```
