```python
import pandas as pd
```

```python
files = {
    'transactions': 'data/transactions_train.csv',
    'articles': 'data/articles.csv',
    'customers': 'data/customers.csv'
}
```

```python
for name, path in files.items():
    df = pd.read_csv(path, nrows = 5)
    print('=' * 55)
    print(name, '->', path)
    print('列名:', list(df.columns))
    print(df.head(), '\n')
```
